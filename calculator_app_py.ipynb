```
# Install widgets support
!pip install ipywidgets -q

from IPython.display import display
import ipywidgets as widgets

# Calculator display
display_box = widgets.Text(
    value='',
    placeholder='0',
    description='',
    layout=widgets.Layout(width='320px', height='50px')
)

# Button style
buttons = [
    ['AC','%','/','*'],
    ['7','8','9','-'],
    ['4','5','6','+'],
    ['1','2','3','='],
    ['0','.']
]

output = widgets.Output()

# Button click function
def press(b):
    key = b.description

    if key == 'AC':
        display_box.value = ''

    elif key == '=':
        try:
            display_box.value = str(eval(display_box.value))
        except:
            display_box.value = "Error"

    elif key == '%':
        try:
            display_box.value = str(float(display_box.value)/100)
        except:
            display_box.value = "Error"

    else:
        display_box.value += key


# Create buttons
rows=[]

for row in buttons:
    btn_row=[]
    for item in row:
        btn = widgets.Button(
            description=item,
            layout=widgets.Layout(width='70px', height='50px')
        )
        btn.on_click(press)
        btn_row.append(btn)

    rows.append(widgets.HBox(btn_row))


# Show calculator
app = widgets.VBox(
    [display_box] + rows,
    layout=widgets.Layout(
        align_items='center'
    )
)

display(app)
```
