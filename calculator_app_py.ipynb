```
import ipywidgets as widgets
from IPython.display import display, clear_output

# Calculator state
display_box = widgets.Text(
    value='',
    placeholder='0',
    description='',
    layout=widgets.Layout(width='320px', height='50px')
)

# Button layout
button_labels = [
    ['AC', '%', '/', '*'],
    ['7', '8', '9', '-'],
    ['4', '5', '6', '+'],
    ['1', '2', '3', '='],
    ['0', '.']
]

# Interaction logic
def on_button_click(b):
    val = b.description
    if val == 'AC':
        display_box.value = ''
    elif val == '=':
        try:
            display_box.value = str(eval(display_box.value))
        except Exception:
            display_box.value = 'Error'
    elif val == '%':
        try:
            display_box.value = str(float(display_box.value) / 100)
        except Exception:
            display_box.value = 'Error'
    else:
        display_box.value += val

# UI Construction
rows = []
for row in button_labels:
    btn_row = []
    for label in row:
        btn = widgets.Button(description=label, layout=widgets.Layout(width='70px', height='50px'))
        btn.on_click(on_button_click)
        btn_row.append(btn)
    rows.append(widgets.HBox(btn_row))

calculator = widgets.VBox([display_box] + rows, layout=widgets.Layout(align_items='center'))

# Clear previous output and display widget to help fix metadata errors
clear_output()
display(calculator)
```
